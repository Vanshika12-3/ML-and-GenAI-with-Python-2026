```python
print("Sum of first 10 natural numbers")
sum = 0;
for i in range(1,11):
  sum += i;
print(sum)
```

```python
print("Factorial of a number")
fact = 1;
num = int(input("Enter your number:"))
for i in range(1,num+1):
  fact *= i;
print(fact)
```

```python
print("Fibonacci series")
a = 0;
b = 1;
n = int(input("Enter the number of terms:"))
print(a)
print(b)
for i in range(n-2):
  c = a+b;
  print(c)
  a=b;
  b=c;
```
